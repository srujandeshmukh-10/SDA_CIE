````python
df = df.dropna(subset=['Appliances'])
print("total missing values are :",df.isnull().sum().sum())
````

````python
df['date'] = pd.to_datetime(df['date'])
df['date'].dtype
````

````python
#creating a new column hour
df['hour'] = df['date'].dt.hour
````

````python
#Phase 3: Exploratory Data Analysis (EDA).
plt.hist(df['Appliances'], bins=30)
plt.xlabel('Appliance Energy Consumption (Wh)')
plt.ylabel('Frequency')
plt.title('Distribution of Appliance Energy Consumption')
plt.show()
````

````python
plt.boxplot(df['Appliances'])
plt.title('Boxplot of Appliance Energy Consumption')
plt.ylabel('Energy Consumption (Wh)')
plt.show()
````

````python
df_c = df.copy()
# Calculate Q1 and Q3
Q1 = df['Appliances'].quantile(0.25)
Q3 = df['Appliances'].quantile(0.75)

# Calculate the Interquartile Range
IQR = Q3 - Q1

# Calculate outlier limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Count outliers
outliers = df[
    (df['Appliances'] < lower_limit) |
    (df['Appliances'] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of outliers:", len(outliers))
print("Percentage of outliers:", len(outliers) / len(df) * 100, "%")
````

````python
# Replace values below the lower limit
df.loc[df['Appliances'] < lower_limit, 'Appliances'] = lower_limit

# Replace values above the upper limit
df.loc[df['Appliances'] > upper_limit, 'Appliances'] = upper_limit

# Check the updated minimum and maximum
print("Minimum:", df['Appliances'].min())
print("Maximum:", df['Appliances'].max())
````

````python
# Count values outside the IQR limits
print("Lower outliers:", (df['Appliances'] < lower_limit).sum())
print("Upper outliers:", (df['Appliances'] > upper_limit).sum())

# Plot the updated boxplot
plt.boxplot(df['Appliances'])
plt.title('Boxplot After IQR Capping')
plt.ylabel('Energy Consumption (Wh)')
plt.show()
````

````python
# Calculate average consumption for each hour
hourly_avg = df.groupby('hour')['Appliances'].mean()

# Plot hourly average consumption
plt.figure(figsize=(10, 5))
plt.plot(hourly_avg.index, hourly_avg.values, marker='o')

plt.xlabel('Hour of the Day')
plt.ylabel('Average Energy Consumption (Wh)')
plt.title('Average Appliance Energy Consumption by Hour')
plt.xticks(range(24))
plt.grid(True)
plt.show()
````

````python
df['day_of_week'] = df['date'].dt.dayofweek
day_names = {
    0: 'Monday',
    1: 'Tuesday',
    2: 'Wednesday',
    3: 'Thursday',
    4: 'Friday',
    5: 'Saturday',
    6: 'Sunday'
}

daily_comparison = df.groupby('day_of_week')['Appliances'].mean()
daily_comparison.index = daily_comparison.index.map(day_names)

plt.figure(figsize=(10, 5))
daily_comparison.plot(kind='bar', edgecolor='black')

plt.xlabel('Day of the Week')
plt.ylabel('Average Appliance Consumption (Wh)')
plt.title('Average Appliance Consumption by Day of the Week')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

print(daily_comparison.round(2))
````

````python
# Explore typical and high appliance consumption levels

percentiles = df['Appliances'].quantile([0.50, 0.75, 0.90, 0.95, 0.99])

print("Appliance consumption percentiles:")
print(percentiles)

print("\nMaximum appliance consumption:")
print(df['Appliances'].max())
````

### EDA Conclusion

Exploratory Data Analysis was performed to understand the distribution, variability, and usage patterns of appliance energy consumption.

The histogram showed that appliance consumption was concentrated at lower values, with a tail extending toward higher consumption values. The initial boxplot identified numerous potential outliers. Using the IQR method, values above the upper limit of 175 Wh were capped at 175 Wh rather than removed. A total of 2,138 observations were reported as exceeding this limit. The subsequent boxplot and percentile analysis showed the effect of this treatment, with the 90th, 95th, and 99th percentiles and the maximum all equal to 175 Wh.

The hourly analysis was used to investigate variations in average appliance consumption throughout the day. The day-of-week analysis also revealed variation in average consumption: Saturday had the highest average at 83.68 Wh, while Tuesday had the lowest at 73.12 Wh, a difference of 10.56 Wh.

Overall, EDA helped identify the distribution of appliance consumption, the impact of outlier capping, and variations in average consumption across time periods. These findings are descriptive and do not establish statistically significant relationships between appliance consumption and other factors. Further descriptive statistics, hypothesis testing, correlation analysis, and regression modelling will be used in subsequent phases to investigate the research questions.

### 3. Outdoor Environmental Variables

### Overall Conclusion

Descriptive statistical analysis established the typical appliance energy consumption, its variability, the shape of its distribution, and the general characteristics of the indoor and outdoor environmental variables. Appliance consumption has a positively skewed distribution, with the mean exceeding both the median and mode. The environmental summaries also show variation in temperature and humidity across measurement locations and over time.
